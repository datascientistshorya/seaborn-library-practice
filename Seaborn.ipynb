### BIVARIATE ANALYSIS

```python
import matplotlib.pyplot as plt
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
plt.title('Tips vs Bill')
plt.show()
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips)
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='size', data=tips)
```

```python
plt.grid(True,color='green',linestyle='dotted')
sns.scatterplot(x='total_bill', y='tip', hue='size', data=tips, size='size')
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='size', data=tips, size='size',style='sex')
```

### Barplot vs Countplot
#### Barplot - shows AVG per category by default black lines= Confidence Interval

```python
sns.barplot(x='day', y='total_bill', data=tips)
```

#### Count Plot -Simple count

```python
sns.countplot(x='day', data=tips)
```

### Stripplot & Swarmplot

```python
sns.stripplot(x='day', y='total_bill', data=tips)
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips)
```

#### Jointplot - Combination of histogram & scatterplot
