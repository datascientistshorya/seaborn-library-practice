```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
titanic=sns.load_dataset('titanic')
iris=sns.load_dataset('iris')
penguins=sns.load_dataset('penguins')
sns.set_style("whitegrid")
```

### LEVEL 1 — Core Seaborn Basics (1–10)

Plot a histogram of total_bill with:

Custom color**

```python
tips.head()
```

```python
sns.histplot(tips['total_bill'],bins=20,kde=True);
```

##### Create a boxplot of total_bill grouped by day.

```python
sns.boxplot(y='total_bill',x='day',data=tips)
```

##### Plot a violin plot of total_bill vs day, split by sex.

```python
plt.figure(figsize=(10,10))
sns.violinplot(y='day',x='total_bill',hue='sex',data=tips)
plt.show()
```

##### Create a scatter plot of total_bill vs tip:Color by sex,Size by size

```python
#sns.scatterplot(x='total_bill',y='tip',color='sex',size='size',data=tips)
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips, size='size')
```

##### Draw a countplot of day ordered by frequency.

```python
day_order = tips['day'].value_counts().index
print(day_order)
print(type(day_order))
sns.countplot(x='day',data=tips,order=day_order)
```

##### Plot a barplot showing average tip per day.

```python
sns.barplot(y='tip',x='day',data=tips)
```

##### sum of tips per day

```python
  sns.barplot(y='tip',x='day',data=tips,estimator=sum)
```

##### Create a strip plot of total_bill vs day with jitter.

```python
sns.stripplot(x='day',y='total_bill',data=tips)
```

##### Convert the strip plot into a swarm plot. What problem does swarmplot solve?

```python
sns.swarmplot(x='day',y='total_bill',data=tips)
```

##### Create a KDE plot for total_bill for: Male vs Female customers

```python
sns.kdeplot(x='total_bill',hue='sex',data=tips,fill=True)
```

##### Use sns.displot() to show: Histogram + KDE Faceted by sex

```python

sns.displot(
    data=tips,
    x="total_bill",
    hue="sex",
    col="sex",
    kde=True,
    bins=20
)

plt.show()
```

##### Rotate x-axis labels in a categorical plot to avoid overlap.

```python
sns.barplot(x='day',y='total_bill',palette='Set2',data=tips)
plt.xticks(rotation=45, fontsize=15)
plt.xlabel("Day of Week", color='red', fontsize=15)
plt.show()
```

##### Add:Title X label Y label

```python
sns.barplot(x='day',y='total_bill',palette='Set2',data=tips)
plt.xticks(rotation=45, fontsize=15)
plt.xlabel("Day of Week", color='red', fontsize=15)
plt.ylabel('Total Bill', color='red',fontsize=15)
plt.title('Total Bill vs Days', color='red',fontsize=20)
plt.show()
```

##### Control axis limits manually on a scatter plot.

```python
sns.set_style('whitegrid')
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.xlim(10, 30)   # x-axis from 10 to 50
plt.ylim(1, 6)  
plt.show()
```

##### Create a plot where:Grid is ON Only y-axis grid is visible


```python
sns.set_style('white')
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.xlim(10, 30)   # x-axis from 10 to 50
plt.ylim(1, 6)  
plt.grid(True,axis='y')
plt.show()
```

##### Create a catplot combining: Boxplot Faceted by sex

```python
sns.catplot(
    x='day',           # categorical variable
    y='total_bill',    # numerical variable
    col='sex',         # facet by sex
    data=tips,
    kind='box')
```

```python
sns.catplot(
    x='day',           # categorical variable
    y='total_bill',    # numerical variable
    col='sex',         # facet by sex
    data=tips,
    kind='violin')
```
